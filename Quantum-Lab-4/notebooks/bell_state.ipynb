# Bell State - Introduction to Qiskit

First introduction to Qiskit. We create a Bell state and run it on the AerSimulator.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
```

```python
# Create the Bell state circuit
# Bell state = (|00> + |11>) / sqrt(2)
qc = QuantumCircuit(2, 2)
qc.h(0)       # Hadamard on qubit 0
qc.cx(0, 1)   # CNOT from qubit 0 to qubit 1
qc.measure([0, 1], [0, 1])
qc.draw('mpl')
```

```python
# Run on the AerSimulator with 2048 shots
sim = AerSimulator()
result = sim.run(qc, shots=2048).result()
counts = result.get_counts()
print(f"Counts : {counts}")
```

```python
# Plot histogram
plot_histogram(counts)
```

We get approximately 50% |00> and 50% |11>, which is the expected Bell state behavior. The two qubits are always measured in the same state (both 0 or both 1) because they are entangled.
